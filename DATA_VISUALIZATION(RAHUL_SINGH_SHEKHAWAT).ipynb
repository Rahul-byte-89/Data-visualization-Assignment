```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df['flavor_profile'] = df['flavor_profile'].astype(str).str.strip().str.lower()
valid = df[~df['flavor_profile'].isin(['-1', 'na', 'none', 'nan', ''])]
flavor_counts = valid['flavor_profile'].value_counts().sort_values(ascending=False)
```

```python
min_share = 0.03
share = flavor_counts / flavor_counts.sum()
major = share[share >= min_share].index
if (share < min_share).any():
    other_sum = flavor_counts[share < min_share].sum()
    flavor_counts_plot = flavor_counts[major].copy()
    if other_sum > 0:
        flavor_counts_plot.loc['other'] = other_sum
else:
    flavor_counts_plot = flavor_counts.copy()
```

```python
plt.figure(figsize=(7.5, 7.5))
colors = plt.cm.Set3(range(len(flavor_counts_plot)))
plt.pie(
    flavor_counts_plot.values,
    labels=[lbl.title() for lbl in flavor_counts_plot.index],
    autopct='%1.1f%%',
    startangle=140,
    counterclock=False,
    colors=colors,
    pctdistance=0.8,
    textprops={'fontsize': 11}
)
plt.title("Most Preferred Flavors (by number of dishes)")
plt.tight_layout()
plt.show()
```

```python
#e.
course_counts = valid['course'].value_counts().sort_values(ascending=False)
plt.figure(figsize=(10,5))
bars = plt.bar(range(len(course_counts)), course_counts.values, color="#1f77b4")
plt.xticks(range(len(course_counts)), [c.title() for c in course_counts.index], rotation=15)
plt.ylabel("Count of Dishes")
plt.xlabel("Course")
plt.title("Most Requested Course (by number of dishes)")
```

```python
#f.
df['Diet'] = df['Diet'].astype(str).str.strip().str.lower()
df['region'] = df['region'].astype(str).str.strip().str.title()
valid = df[~df['region'].isin(['-1', 'Na', 'None', 'Nan', ''])]
valid = valid[valid['Diet'].isin(['vegetarian', 'non vegetarian'])]
counts = valid.pivot_table(index='region', columns='Diet', values='Name', aggfunc='count', fill_value=0)
```

```python
if 'vegetarian' not in counts.columns:
    counts['vegetarian'] = 0
if 'non vegetarian' not in counts.columns:
    counts['non vegetarian'] = 0
```

```python
counts = counts.assign(total=counts.sum(axis=1)).sort_values('total', ascending=False).drop(columns='total')
```

```python
regions = counts.index.tolist()
veg = counts['vegetarian'].to_numpy()
nonveg = counts['non vegetarian'].to_numpy()
```

```python
x = np.arange(len(regions))
width = 0.38

plt.figure(figsize=(11,5))
bars1 = plt.bar(x - width/2, veg, width, label='Vegetarian', color='#2ca02c')
bars2 = plt.bar(x + width/2, nonveg, width, label='Non-Vegetarian', color='#d62728')

plt.xticks(x, regions, rotation=20)
plt.ylabel('Number of Orders')
plt.title('Vegetarian vs Non-Vegetarian Orders by Region (Grouped)')
plt.legend()

for b in list(bars1) + list(bars2):
    h = b.get_height()
    plt.text(b.get_x() + b.get_width()/2, h, f'{int(h)}', ha='center', va='bottom', fontsize=9)

```
